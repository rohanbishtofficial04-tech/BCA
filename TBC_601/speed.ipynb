```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

speed=np.array([4.0,4.5,5.0,5.5,6.0,6.5,7.0])

mean_value=np.mean(speed)
print("Mean Speed: ",mean_value)
median_value=np.median(speed)
print("Median Speed: ",median_value)
mode_value=stats.mode(speed)
print("Mode Speed: ",mode_value[0])
std_deviation=np.std(speed)
print("Standard Deviation: ",std_deviation)
variance_value=np.var(speed)
print("Variance: ",variance_value)
print("\n")
plt.boxplot(speed)
plt.title("Boxplot of Speed")
plt.ylabel("Speed")
plt.show()

def remove_outliers(data):
    Q1=np.percentile(data,25)
    Q3=np.percentile(data,75)
    IQR=Q3-Q1
    lower_bound=Q1-1.5*IQR
    upper_bound=Q3+1.5*IQR
    filtered_data=data[(data>=lower_bound) & (data<=upper_bound)]
    return filtered_data
print("\n")
cleaned_speed=remove_outliers(speed)
print("Cleaned Speed: ",cleaned_speed)
cleaned_mean=np.mean(cleaned_speed)
print("Cleaned Mean Speed: ",cleaned_mean)
cleaned_median=np.median(cleaned_speed)
print("Cleaned Median Speed: ",cleaned_median)
cleaned_mode=stats.mode(cleaned_speed)
print("Cleaned Mode Speed: ",cleaned_mode[0])
```
